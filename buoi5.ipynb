```python
#W5A1
def max_of_two(a,b)->int:
    return max(a,b)
a,b=map(int,input().split())
print(max_of_two(a,b))
```

```python
#W5A2
def swap(a,b)->int:
    a,b=b,a
    return a,b
a,b=map(int,input().split())
print(swap(a,b))
```

```python
#W5A3
def check_prime(n):
    if n<2:
        return False
    for i in range(2,n):
        if n%i==0:
            return False
        else:
            return True
n=int(input())
print(check_prime(n))
```

```python
#W5A4
def check_perfect(num):
    tmp=0
    for i in range(1,(num//2)+1):
        if num%i==0:
            tmp+=i
    if tmp==num:
        return True
    else:
        return False
n=int(input())
print(check_perfect(n))
```

```python
#W5A6
def giaithua(n):
    res=1
    for i in range(1,n+1):
        res*=i
    return res
n=int(input())
print(giaithua(n))
```

```python
#W5A7
def calculate(a,b,k):
    if k=='+':
        return a+b
    elif k=='-':
        return a-b
    elif k=='*':
        return a*b
    elif k=='/':
        return a/b
a,k,b=map(str,input().split())
a=float(a)
b=float(b)
print(f"{calculate(a, b, k):.2f}")
```

```python
#W5A8
def binary(n):
    ans = ""
    while n > 0:
        ans = str(n % 2) + ans
        n //= 2
    return ans
def hamming_distance(x,y):
    if len(x) < len(y):
        x = '0' * (len(y) - len(x)) + x
    elif len(y) < len(x):
        y = '0' * (len(x) - len(y)) + y
    cnt=0
    for i in range(len(x)):
        if x[i] != y[i]:
            cnt += 1
    return cnt
a,b = map(int,input().split())
x=binary(a)
y=binary(b)
print(hamming_distance(x,y))
```

```python
#W5A9
def tongcs(n):
    res=0
    while n>0:
        res+=n%10
        n//=10
    return res
n=int(input())
print(tongcs(n))
```

```python
#W5A10
def check(s1,s2):
    if len(s1) != len(s2):
        return False
    a=[s1[1]]
    b=[s2[1]]
    k=[0]
    cnt=1
    for i in range(1,len(s1)):
        if s1[i] in a:
            for j in range(cnt):
                if a[j] == s1[i]:
                    if b[j] != s2[i]:
                        return False
        elif s2[i] in b:
            for j in range(cnt):
                if b[j] == s2[i]:
                    if b[j] != s1[i]:
                        return False
        elif s1[i] not in a and s2[i] not in b:
            a.append(s1[i])
            b.append(s2[i])
            cnt+=1
    return True
a,b=map(str,input().split())
print(check(a,b))
```
