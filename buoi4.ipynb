```python
#W4A7
import math
def check_prime(n):
    if(n<2):
        return False
    for i in range(2, int(math.sqrt(n))+1):
        if n % i == 0:
            return False
    return True
def factor_prime(n):
    for i in range(n, 1, -1):
        if n % i == 0:
            if (check_prime(i)):
                return i
n=int(input())
print(factor_prime(n))
```

```python
#W4A8
def reverse(n):
    s=str(n)
    return int(s[::-1])
def check_palindrome(n):
    s=str(n)
    for i in range(int(len(s)/2)):
        if s[i] != s[len(s)-i-1]:
            return False
    return True
n=int(input())
ans=1
n+=reverse(n)
while check_palindrome(n)==False:
    n+=reverse(n)
    ans+=1
print(ans,n)


```

```python
#W4A9
import math
def check_perfect_square(x,n):
    if x>n or x==0:
        return False
    s=str(x)
    a=[0]*10
    for i in s:
        a[int(i)]+=1
        if a[int(i)]>1:
            return False
    return True

n=int(input())
cuoi=int(math.sqrt(n))+1
ans=""
for i in range(cuoi+1):
    if check_perfect_square(i*i,n):
        ans=ans+str(i*i)+" "
print(ans)
```

```python
#W4A10
def collatz(n):
    ans=1
    while n>1:
        if n%2==0:
            n/=2
            ans+=1
        elif n%2!=0:
            n=n*3+1
            ans+=1
    return ans

n=int(input())
x=0
y=0
for i in range(1,n+1):
    tmp=collatz(i)
    if tmp>y:
        x=i
        y=tmp
print(x,y)
```
