```python
#W11A3
def happy_number(n, seen):
    ans = sum(int(i) ** 2 for i in str(n))
    if ans == 1:
        return True
    if ans in seen:
        return False
    seen.add(ans)
    return happy_number(ans, seen)


try:
    inp = input("Nhập vào một số nguyên dương a: ")
    a = int(inp)
    if happy_number(a, set()):
        print("YES")
    else:
        print("NO")
except ValueError:
    print("Vui lòng nhập số nguyên hợp lệ!")
```

```python
#W11A5
a = list(map(str, input().split()))
ans=[]
for i in range(len(a)):
    try:
        a[i] = int(a[i])
        ans.append(a[i])
    except ValueError:
        print(f"'{a[i]}' is not a number.")
print(f"Danh sách sau khi chuyển là: {ans}")
```
