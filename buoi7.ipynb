```python
#W7A1
a=list(map(int,input().split()))
def binary_search(a,left,right,k):
    low=left
    high=right
    while low<=high:
        mid=(low+high)//2
        if a[mid]==k:
            return mid
        elif a[mid]<k:
            high=mid-1
        elif a[mid]>k:
            low=mid+1
    return -1
k=int(input())
print(binary_search(a,0,len(a)-1,k))
```

```python
#W7A3
def bubble_sort(arr: list[int]) -> list[int]:
    n = len(arr)
    swap = False
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swap = True
        if not swap:
            return arr
    return arr
def selection_sort(arr: list[int]) -> list[int]:
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i+1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr
lst = list(map(int,input().split()))
print(bubble_sort(lst))
```

```python
#W7A10
nums = eval(input())
m = int(input())
max_sum = 0
curr = 0
prefixes = [0]
for i in nums:
  curr = (curr + i) % m
  for prev in prefixes:
    temp_sum = (curr - prev) % m
    if temp_sum > max_sum:
      max_sum = temp_sum
  prefixes.append(curr)
print(max_sum)
```
