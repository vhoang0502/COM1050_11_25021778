```python
#W6A3
a=tuple(map(int,input().split()))
a=list(a)
k=int(input())
k=k%len(a)
p1=a
p2=[]
while k>0:
    p2.append(p1[0])
    p1=p1[1:]
    k-=1
p1=tuple(p1)
p2=tuple(p2)
print(p1+p2)
```

```python
#W6A4
a=list(map(str,input().split()))
my_dict={}
for i in a:
    b=i.split(":")
    key=b[0]
    value=b[1]
    if key not in my_dict:
        my_dict[key]=list(value)
    elif key in my_dict:
        my_dict[key].extend(value)
print(my_dict)
```

```python
#W6A10
a=list(map(int,input().split()))
k=int(input())
ans=[]
x=[0,0]
for i in range(len(a)-1):
    for j in range(i+1,len(a)):
        if a[i]+a[j]==k:
            x[0]=a[i]
            x[1]=a[j]
            y=tuple(x)
            if y not in ans:
                ans.append(y)
ans.sort()
print(ans)
```

```python
#W6A12
a=list(map(int,input().split()))
d={}
for i in range(0,len(a)):
    if a[i] not in d:
        d[a[i]]=1
    else:
        d[a[i]]+=1
ans=0
count=0
for i in d.keys():
    if d[i]>count:
        count=d[i]
        ans=i
    elif d[i]==count:
        if i<ans:
            ans=i
print(ans)
```

```python
#W6A20
a=list(map(int,input().split()))
k=int(input())
def tk(a,k):
    ans=-1
    for i in range(len(a)):
        if a[i]==k:
            return i+1
    return ans
print(tk(a,k))
```
