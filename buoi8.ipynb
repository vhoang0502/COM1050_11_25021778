```python
#W8A1
class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
    def cal_s_tp(self,radius,height):
        s = radius*height*2*3.14+4*radius*3.14
        return s
    def cal_v(self,radius,height):
        v=height*radius*radius*3.14
        return v
r,h=map(float,input().split())
Cylinder = Cylinder(r,h)
ans1=Cylinder.cal_s_tp(r,h)
ans2=Cylinder.cal_v(r,h)
print(f"{ans1:.2f} {ans2:.2f}")
```

```python
#W8A2
class Date:
    def __init__(self, day: int, month: int, year: int):
        self.day = day
        self.month = month
        self.year = year
    def is_leap_year(self):
        return self.year % 4 == 0 and self.year % 100 != 0 or self.year % 400 == 0
    def check_date(self):
        if self.month<1 or self.month>12:
            return False
        days_in_month = [ 31, 28+self.is_leap_year(), 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        if self.day<1 or self.day>days_in_month[self.month-1]:
            return False
        return True
    def next_day(self):
        days_in_month = [31, 28 + self.is_leap_year(), 31, 30, 31, 30,31, 31, 30, 31, 30, 31]
        self.day += 1
        if self.day > days_in_month[self.month - 1]:
            self.day = 1
            self.month += 1
            if self.month > 12:
                self.month = 1
                self.year += 1


day,month,year = map(int,input().split("/"))
date=Date(day,month,year)
if not date.check_date():
    print("INVALID")
else:
    date.next_day()
    print(f"{date.day:02d}/{date.month:02d}/{date.year}")
```

```python
#W8A3
import math
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def is_gtd(self):
        return self.x == 0 and self.y == 0
    def is_x(self):
        return self.y == 0
    def is_y(self):
        return self.x == 0
    def distance(self):
        return math.sqrt(self.x**2 + self.y**2)
x,y=map(float,input().split())
point=Point(x,y)
dis=point.distance()
if point.is_gtd():
    print("M là gốc tọa độ")
    if point.is_x():
        print("M nằm trên trục hoành")
    elif point.is_y():
        print("M nằm trên trục tung")
    print(f"Khoảng cách từ M đến gốc tọa độ là {dis:.2f}")
else:
    print("M không phải gốc tọa độ")
    if point.is_x():
        print("M nằm trên trục hoành")
    elif point.is_y():
        print("M nằm trên trục tung")
    print(f"Khoảng cách từ M đến gốc tọa độ là {dis:.2f}")
```

```python
#W8A4
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b

    def add(self):
        return self.a + self.b

    def subtract(self):
        return self.a - self.b

    def product(self):
        return self.a * self.b

    def divide(self):
        if self.b == 0:
            return "ERROR"
        return self.a / self.b

    def power(self):
        return self.a ** self.b

    def mod(self):
        if self.b == 0:
            return "ERROR"
        return self.a % self.b

    def set_numbers(self, x, y):
        self.a = x
        self.b = y


def main():
    a, b = map(float, input("Nhập hai số ban đầu: ").split())
    cal = Calculator(a, b)

    while True:
        print(f"\nHai số hiện tại: {cal.a} và {cal.b}")
        print("___________ MENU ___________")
        print("1. Add        2. Subtract")
        print("3. Product    4. Divide")
        print("5. Power      6. Mod")
        print("7. Set Numbers")
        print("-----------------------------")

        t = int(input("Nhập lựa chọn: "))

        if t == 1:
            print(cal.add())
        elif t == 2:
            print(cal.subtract())
        elif t == 3:
            print(cal.product())
        elif t == 4:
            print(cal.divide())
        elif t == 5:
            print(cal.power())
        elif t == 6:
            print(cal.mod())
        elif t == 7:
            x, y = map(float, input("Nhập hai số mới: ").split())
            cal.set_numbers(x, y)
        else:
            print("Lựa chọn không hợp lệ")

        ok = input("Bạn có muốn thoát? ").strip().lower()
        if ok == "yes":
            break


main()

```

```python
#W8A5
class ShoppingCart:
    def __init__(self):
        self.cart_items = []

    def add_item(self, name, price):
        self.cart_items.append({"name": name, "price": price})

    def remove_item(self, name):
        for item in self.cart_items:
            if item["name"] == name:
                self.cart_items.remove(item)
                return True
        return False

    def is_empty(self):
        return len(self.cart_items) == 0

    def total_price(self):
        total = 0
        for item in self.cart_items:
            total += item["price"]
        return total

    def display(self):
        if self.is_empty():
            print("Giỏ hàng đang rỗng.")
            return
        for item in self.cart_items:
            print(f"Tên: {item['name']} | Giá: {item['price']}")

    def clear_cart(self):
        self.cart_items.clear()


def input_int(prompt):
    while True:
        try:
            return int(input(prompt))
        except ValueError:
            print("Vui lòng nhập số.")


def main():
    cart = ShoppingCart()
    print("Welcome to Shopping Cart")

    while True:
        print("\n---------------- MENU ----------------")
        print("2. Add item")
        print("3. Remove item")
        print("4. Check empty")
        print("5. Total price")
        print("6. Display cart")
        print("7. Clear cart")
        print("-------------------------------------")

        choice = input_int("Your choice: ")

        if choice == 2:
            name = input("Enter item name: ")
            price = float(input("Enter item price: "))
            cart.add_item(name, price)
            print("Item added.")

        elif choice == 3:
            name = input("Enter item name to remove: ")
            if cart.remove_item(name):
                print("Item removed.")
            else:
                print("Không tìm thấy sản phẩm.")

        elif choice == 4:
            print("Giỏ hàng rỗng." if cart.is_empty() else "Giỏ hàng có sản phẩm.")

        elif choice == 5:
            print("Tổng tiền:", cart.total_price())

        elif choice == 6:
            cart.display()

        elif choice == 7:
            cart.clear_cart()
            print("Đã xóa toàn bộ giỏ hàng.")

        ans = input("Có muốn thoát chương trình không? ").strip().lower()
        if ans == "yes":
            break


main()
```

```python
#W8A6
class Rectangle:
    def __init__(self, width, height ,k):
        self.w = width
        self.h = height
        self.k = k
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self):
        self.w = self.w * self.k
        self.h = self.h * self.k
w,h,k=map(float,input().split())
rec=Rectangle(w, h,k)
rec.scale()
print(f"{rec.area():.2f} {rec.perimeter():.2f}")
```

```python
#W8A7
import math

class Fraction:
    def __init__(self, num, den):
        if den == 0:
            raise ValueError("Denominator cannot be zero")
        if den < 0:
            num = -num
            den = -den

        g = math.gcd(num, den)
        self.num = num // g
        self.den = den // g

    def __str__(self):
        return f"{self.num}/{self.den}"

    def add(self, other):
        num = self.num * other.den + other.num * self.den
        den = self.den * other.den
        return Fraction(num, den)

    def sub(self, other):
        num = self.num * other.den - other.num * self.den
        den = self.den * other.den
        return Fraction(num, den)

    def mul(self, other):
        num = self.num * other.num
        den = self.den * other.den
        return Fraction(num, den)

    def div(self, other):
        if other.num == 0:
            raise ValueError("Cannot divide by zero")
        num = self.num * other.den
        den = self.den * other.num
        return Fraction(num, den)

a, b, op, c, d = input().split()

f1 = Fraction(int(a), int(b))
f2 = Fraction(int(c), int(d))

if op == '+':
    print(f1.add(f2))
elif op == '-':
    print(f1.sub(f2))
elif op == '*':
    print(f1.mul(f2))
elif op == '/':
    print(f1.div(f2))
```

```python
#W8A8
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount < self.balance:
            self.balance -= amount

    def __str__(self):
        return f"{self.owner} : {self.balance}"

x,y=input().split()
initial_balance = float(y)

n=int(input())
bank = BankAccount(x,initial_balance)
while n > 0:
    a,b=input().split()
    b=float(b)
    if a=="WITHDRAW":
        bank.withdraw(b)
    elif a=="DEPOSIT":
        bank.deposit(b)
    n-=1
print(bank)
```

```python
#W8A9
class Student:
    def __init__(self, name,number):
        self.name = name
        self.number = number
        self.score=0

    def add_score(self,hm):
        self.score+=hm

    def avg(self):
        temp= self.score/self.number
        if temp>=8.0:
            return "Excellent"
        elif (temp>=6.5) and (temp<8.0):
            return "Good"
        elif (temp>=5) and (temp<6.5):
            return "Average"
        else:
            return "Poor"

    def __str__(self):
        return f"{self.name} {(self.score/self.number):.2f} {self.avg()}"

stu=input()
n=int(input())
student=Student(stu,n)
for i in range(n):
    x,y=input().split()
    student.add_score(float(y))
print(student)
```

```python
#W8A10
class Book:
    def __init__(self, title, author , year):
        self.title = title
        self.author = author
        self.year = year

class Library:
    def __init__(self, books):
        self.books = books

    def add_book(self, book):
        self.books.append(book)

    def count_by_author(self, author):
        count = 0
        for book in self.books:
            if book.author == author:
                count += 1
        return count

    def count_by_year(self, year):
        count = 0
        for book in self.books:
            if book.year == year:
                count += 1
        return count

n=int(input())
lib=Library([])
while n>0:
    x,y=input().split()
    if x=="ADD":
        title, author, year = y.split(";")
        lib.add_book(Book(title, author, year))
    elif x=="COUNT":
        print(lib.count_by_author(y))
    else:
        print(lib.count_by_year(y))
    n-=1
```

```python
#W8A11
import math

def vector_sub(a, b):
    return (a[0]-b[0], a[1]-b[1], a[2]-b[2])

def cross(u, v):
    return (
        u[1]*v[2] - u[2]*v[1],
        u[2]*v[0] - u[0]*v[2],
        u[0]*v[1] - u[1]*v[0]
    )

def dot(u, v):
    return u[0]*v[0] + u[1]*v[1] + u[2]*v[2]

def norm(u):
    return math.sqrt(dot(u, u))

A = tuple(map(float, input().split()))
B = tuple(map(float, input().split()))
C = tuple(map(float, input().split()))
D = tuple(map(float, input().split()))

AB = vector_sub(B, A)
AC = vector_sub(C, A)
n1 = cross(AB, AC)

BC = vector_sub(C, B)
BD = vector_sub(D, B)
n2 = cross(BC, BD)

cos_phi = abs(dot(n1, n2)) / (norm(n1) * norm(n2))
phi = math.degrees(math.acos(cos_phi))

print(f"{phi:.2f}")
```

```python
#W8A12
class ComplexNumber:
    def __init__(self, realPart, imaginaryPart):
        self.realPart = realPart
        self.imaginaryPart = imaginaryPart

    def multiply(self, other):
        real = self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart
        imag = self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart
        return ComplexNumber(real, imag)

a, b = map(float, input().split())
c, d = map(float, input().split())

z1 = ComplexNumber(a, b)
z2 = ComplexNumber(c, d)

result = z1.multiply(z2)
print(result.realPart, result.imaginaryPart)
```

```python
#W8A13
import math
class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.edge1 = edge1
        self.edge2 = edge2
        self.edge3 = edge3

    def getArea(self):
        a, b, c = self.edge1, self.edge2, self.edge3

        if a + b <= c or a + c <= b or b + c <= a:
            return -1

        p = (a + b + c) / 2
        return math.sqrt(p * (p - a) * (p - b) * (p - c))

a, b, c = map(int, input().split())
triangle = Triangle(a, b, c)

area = triangle.getArea()

if area == -1:
    print("invalid")
else:
    print(f"{area:.2f}")
```
